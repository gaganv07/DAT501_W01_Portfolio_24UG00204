```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import confusion_matrix, classification_report
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense, Input

df = pd.read_csv("/content/Churn_Modelling.csv")
df = df.drop(columns=["RowNumber", "CustomerId", "Surname"])   # no signal

# Know what beating the trivial model even means, BEFORE training anything.
baseline = df["Exited"].value_counts(normalize=True).max()
print(f"majority-class baseline: {baseline:.1%}")              # 79.6%

df = pd.get_dummies(df, columns=["Geography", "Gender"], drop_first=True)
X = df.drop(columns=["Exited"]).astype(float)                  # 11 features
y = df["Exited"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=1, stratify=y)

scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)    # fit on TRAIN only
X_test = scaler.transform(X_test)          # test is only transformed

model = Sequential([
    Input(shape=(X_train.shape[1],)),
    Dense(11, activation="relu"),
    Dense(11, activation="relu"),
    Dense(1, activation="sigmoid"),        # binary -> 1 node + sigmoid
])
model.compile(optimizer="adam", loss="binary_crossentropy", metrics=["accuracy"])
model.fit(X_train, y_train, epochs=50, batch_size=32, validation_split=0.2)

proba = model.predict(X_test).ravel()      # probabilities, not labels
pred = (proba > 0.5).astype(int)           # 0.5 is a CHOICE, not a rule
print(confusion_matrix(y_test, pred))
print(classification_report(y_test, pred))  # read recall on class 1
```
